```python
from multidata import pose, manifest, casematch
import os
import sqlite3
import pandas as pd

DATA_DIR = '../data/raw'
MANIFEST_PATH = '/Users/kxw680/projects/multidata-local/manifest.sqlite'
```

```python
filename = '20260212-124505-v307721-37.mp4'
case_id = casematch.resolve(os.path.join(DATA_DIR, filename), manifest_path=MANIFEST_PATH)
case_id
camera_id = filename.split('-')[-1].split('.')[0]
```

```python
# 398457 was 254m before pose modification
# 54m after
pose.extract(
    os.path.join(DATA_DIR, filename),
    out_path=os.path.join("../data/pose", f"{case_id}_{camera_id}_motion_only.pkl"),
    device="cpu",
)
```

```python
import json
import numpy as np
data = json.load(open(os.path.join("../data/pose", f'{case_id}_{camera_id}_motion_only.json')))
data["keypoint"] = np.asarray(data["keypoint"], dtype=np.float32)
data["keypoint_score"] = np.asarray(data["keypoint_score"], dtype=np.float32)
```

```python
pose.render_overlay(os.path.join(DATA_DIR, filename),
entry=data, out_path=os.path.join("../data/pose", f'{case_id}_{camera_id}_motion_only_overlay.mp4'),
)
```
