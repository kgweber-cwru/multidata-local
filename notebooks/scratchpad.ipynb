```python
import pandas as pd
import sqlite3
import os
import datetime
import re
from multidata import casematch, ingest, manifest
from multidata.paths import RAW_DIR, MANIFEST_PATH

DATA_DIR = RAW_DIR

filename = '20260226-155705-v309393-20.mp4'
case_id = casematch.resolve(os.path.join(DATA_DIR, filename), manifest_path=MANIFEST_PATH)
case_id
```

```python
for file in os.listdir(DATA_DIR):
    if file[-4:] == ".mp4":
        fpath = os.path.join(DATA_DIR, file)
        try:
            case_id = casematch.resolve(fpath, manifest_path=MANIFEST_PATH)
            info = ingest.summarize(fpath)
            try:
                manifest.add_video(case_id=case_id, path=MANIFEST_PATH, **info)
            except Exception as e:
                print(f"{file}: {e}")
                continue
        except casematch.CaseMatchError as e:
            print(f"{file}: {e}")
```

```python
from multidata import audio

audio.extract(os.path.join(DATA_DIR, filename), 
              out_path=os.path.join("../data/audio", f'{case_id}_audio.wav')
)
```
