```python
import os
import requests
from dotenv import load_dotenv

load_dotenv()
LS_HOST = "https://simlearningspace.case.edu"
API_KEY = os.getenv("LS_API_KEY")

r = requests.post(
    f"{LS_HOST}/w/api/v2/auth.cgi",
    json={"command": "token-for-apikey", "apiKeySecret": API_KEY},
)
r.raise_for_status()
token = r.json()["accessToken"]
```

```python
s = requests.Session()
s.headers["Authorization"] = f"Bearer {token}"

cases = s.get(f"{LS_HOST}/api/data/v1/cases", params={"limit": 3, "offset": 0}).json()
cases["total"], len(cases["data"])
```
