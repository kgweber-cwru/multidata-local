```python
import pandas as pd
import sqlite3

asdf = pd.read_sql("""SELECT a.*, v.duration_s/60 as duration_m, r.case_title
                   FROM assignments a 
                   join videos v on a.recording_id = v.recording_id
                   join recordings r on a.recording_id = r.recording_id""", 
                   sqlite3.connect("../manifest.sqlite"))
```

```python
asdf
```

```python
latest = (
    asdf[asdf.status != 'abandoned']
      .sort_values(['recording_id', 'pass'])
      .drop_duplicates('recording_id', keep='last')
)

latest_gold     = latest[latest.status == 'gold']
latest_not_gold = latest[latest.status != 'gold']
```
