```python
import pandas as pd
import sqlite3
from multidata import manifest, paths
```

```python
conn = sqlite3.connect(paths.MANIFEST_PATH)
```

```python
cases = pd.read_sql_query("""SELECT c.case_id, c.event_name, c.case_name, c.room_number, c.learner_name,
                                v.duration_s / 60 as duration_min, v.filepath, v.asr_status
                                FROM cases c
                                INNER JOIN video_case_matches vcm on c.case_id = vcm.case_id
                                INNER JOIN videos v on vcm.case_id = v.case_id and vcm.camera = v.camera

                                """, conn)
```

```python
cases.sort_values(by="duration_min", ascending=True)
```
