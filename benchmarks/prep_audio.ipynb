```python
import pandas as pd
import sqlite3
from multidata import manifest, paths
```

```python
conn = sqlite3.connect(paths.MANIFEST_PATH)
```

```python
recordings = pd.read_sql_query("""SELECT c.recording_id, c.activity_title, c.case_title, c.room_number, c.learner_name,
                                v.duration_s / 60 as duration_min, v.filepath, v.asr_status
                                FROM recordings c
                                INNER JOIN file_matches vcm on c.recording_id = vcm.recording_id
                                INNER JOIN videos v on vcm.recording_id = v.recording_id and vcm.camera = v.camera

                                """, conn)
```

```python
recordings.sort_values(by="duration_min", ascending=True)
```
